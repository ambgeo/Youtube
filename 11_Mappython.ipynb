```python
##Importar
import geopandas as gpd
import leafmap
```

```python
gdf = gpd.read_file("/content/escolas_filtradas.shp")

# Conferir os dados
gdf.head()
```

```python
gdf = gdf.to_crs(epsg=4326)
```

```python
# Criar mapa
m = leafmap.Map(
    center=[
        gdf.geometry.y.mean(),
        gdf.geometry.x.mean()
    ],
    zoom=10
)

# Adicionar camada de escolas
m.add_gdf(
    gdf,
    layer_name="Escolas",
    info_mode="on_hover"  # mostra atributos ao passar o mouse
)

m
```

```python
# Se tiver geometria não-ponto, usa o centróide
geom = gdf.geometry
if not all(geom.geom_type == "Point"):
    geom = geom.centroid
# Criar lista [lat, lon] (isso é o que o heatmap precisa)
heat_data = [[pt.y, pt.x] for pt in geom if pt is not None]
# Criar mapa
m = leafmap.Map(
    center=[gdf.geometry.y.mean(), gdf.geometry.x.mean()],
    zoom=10
)
# Pontos (opcional)
m.add_gdf(gdf, layer_name="Escolas", info_mode="on_hover")
# Heatmap
m.add_heatmap(
    heat_data,
    name="Heatmap de Escolas",
    radius=25,
    blur=15,
    min_opacity=0.4
)
m
```
