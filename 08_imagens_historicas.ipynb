```python
import leafmap.maplibregl as leafmap
```

Adicione as imagens históricas

```python
m = leafmap.Map(center=[ -20.12, -44.12], zoom=16, sidebar_visible=True)
m.add_wayback_layer(date="2019-01-01")
m.add_wayback_layer(date="2025-09-25")
m
```

Adicione um timeslider

```python
m = leafmap.Map(
    center=[-20.12, -44.12],
    zoom=16,
    sidebar_visible=True,
    layer_manager_expanded=False,
)
m.add_wayback_time_slider(default_index=-1)
m
```
